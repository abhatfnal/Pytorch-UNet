```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

# Base paths for the different samples
base_paths = {
    "nomMC": "/scratch/7DayLifetime/abhat/wirecell/dnn_roi_icarus/training/bnb_cosmics/nomMC",
    "randMC": "/scratch/7DayLifetime/abhat/wirecell/dnn_roi_icarus/training/bnb_cosmics/randMC",
    "opaqueMC": "/scratch/7DayLifetime/abhat/wirecell/dnn_roi_icarus/training/bnb_cosmics/opaqueMC"
}

# Function to read CSV and plot the losses
def plot_losses(plane_path, plane_name, sample_name, pdf):
    # Paths to the CSV files
    train_loss_path = os.path.join(plane_path, "train-loss.csv")
    eval_loss_path = os.path.join(plane_path, "eval-loss.csv")
    eval_dice_path = os.path.join(plane_path, "eval-dice.csv")

    # Load data from CSV files
    train_loss = pd.read_csv(train_loss_path, header=None, names=["Epoch", "Loss"])
    eval_loss = pd.read_csv(eval_loss_path, header=None, names=["Epoch", "Loss"])
    eval_dice = pd.read_csv(eval_dice_path, header=None, names=["Epoch", "Dice Coefficient"])

    # Create side-by-side plots
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Plot training and validation loss with markers
    axes[0].plot(train_loss['Epoch'], train_loss['Loss'], marker='o', label='Training Loss')
    axes[0].plot(eval_loss['Epoch'], eval_loss['Loss'], marker='x', label='Validation Loss')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Loss')
    axes[0].set_title(f'{plane_name} - {sample_name} - Loss')
    axes[0].legend()
    axes[0].grid(True)

    # Plot validation dice with markers
    axes[1].plot(eval_dice['Epoch'], eval_dice['Dice Coefficient'], marker='s', label='Validation Dice')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Dice Coefficient')
    axes[1].set_title(f'{plane_name} - {sample_name} - Dice Coefficient')
    axes[1].legend()
    axes[1].grid(True)
    plt.tight_layout()

    # Save the figure to the PDF
    pdf.savefig(fig)
    plt.close(fig)

# Path to save the final PDF
output_pdf_path = "ICARUS_training.pdf"

# Create a PDF to save all plots
with PdfPages(output_pdf_path) as pdf:
    # Iterate over each sample and each plane
    for sample_name, base_path in base_paths.items():
        u_plane_path = os.path.join(base_path, "U_Plane")
        v_plane_path = os.path.join(base_path, "V_Plane")

        # Plot for U Plane
        plot_losses(u_plane_path, "U Plane", sample_name, pdf)

        # Plot for V Plane
        plot_losses(v_plane_path, "V Plane", sample_name, pdf)

print(f"All plots have been saved to {output_pdf_path}")
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

# Base paths for the different samples
base_paths = {
    "nomMC": "/scratch/7DayLifetime/abhat/wirecell/dnn_roi_icarus/training/mixed_samples/nom",
    "randMC": "/scratch/7DayLifetime/abhat/wirecell/dnn_roi_icarus/training/mixed_samples/rand",
    # "opaqueMC": "/scratch/7DayLifetime/abhat/wirecell/dnn_roi_icarus/training/bnb_cosmics/opaqueMC"
}

# Function to read CSV and plot the losses
def plot_losses(plane_path, plane_name, sample_name, pdf):
    # Paths to the CSV files
    train_loss_path = os.path.join(plane_path, "train-loss.csv")
    eval_loss_path = os.path.join(plane_path, "eval-loss.csv")
    eval_dice_path = os.path.join(plane_path, "eval-dice.csv")

    # Check if all required files exist
    if not all(os.path.exists(path) for path in [train_loss_path, eval_loss_path, eval_dice_path]):
        print(f"Skipping {plane_name} - {sample_name} due to missing CSV files.")
        return

    # Load data from CSV files
    train_loss = pd.read_csv(train_loss_path, header=None, names=["Epoch", "Loss"])
    eval_loss = pd.read_csv(eval_loss_path, header=None, names=["Epoch", "Loss"])
    eval_dice = pd.read_csv(eval_dice_path, header=None, names=["Epoch", "Dice Coefficient"])

    # Create side-by-side plots
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Plot training and validation loss with markers
    axes[0].plot(train_loss['Epoch'], train_loss['Loss'], marker='o', label='Training Loss')
    axes[0].plot(eval_loss['Epoch'], eval_loss['Loss'], marker='x', label='Validation Loss')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Loss')
    axes[0].set_title(f'{plane_name} - {sample_name} - Loss')
    axes[0].legend()
    axes[0].grid(True)

    # Plot validation dice with markers
    axes[1].plot(eval_dice['Epoch'], eval_dice['Dice Coefficient'], marker='s', label='Validation Dice')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Dice Coefficient')
    axes[1].set_title(f'{plane_name} - {sample_name} - Dice Coefficient')
    axes[1].legend()
    axes[1].grid(True)
    plt.tight_layout()

    # Save the figure to the PDF
    pdf.savefig(fig)
    plt.close(fig)

# Path to save the final PDF
output_pdf_path = "ICARUS_training_mixed_SGD.pdf"

# Create a PDF to save all plots
with PdfPages(output_pdf_path) as pdf:
    # Iterate over each sample and each plane
    for sample_name, base_path in base_paths.items():
        u_plane_path = os.path.join(base_path, "U_Plane/SGD")
        v_plane_path = os.path.join(base_path, "V_Plane/SGD")

        # Plot for U Plane
        plot_losses(u_plane_path, "U Plane", sample_name, pdf)

        # Plot for V Plane
        plot_losses(v_plane_path, "V Plane", sample_name, pdf)

print(f"All plots have been saved to {output_pdf_path}")
```
